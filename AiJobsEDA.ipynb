```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.select_dtypes('number').corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

```python
sns.histplot(df["salary_usd"], bins=40, kde=True).set(title="AI Job Salary Distribution", xlabel="Salary (USD)", ylabel="Frequency")
plt.show()
```

```python
sns.barplot(x=df["job_title"].value_counts().head(10), y=df["job_title"].value_counts().head(10).index).set(title="Top 10 AI Job Titles", xlabel="Number of Jobs", ylabel="Job Title")
plt.show()
```

```python
top_countries = df["company_location"].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title("Top Countries Hiring AI Professionals")
plt.xlabel("Number of Jobs")
plt.ylabel("Country")

plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x=df["remote_ratio"])

plt.title("Remote Work Distribution")
plt.xlabel("Remote Ratio (0 = Onsite, 50 = Hybrid, 100 = Fully Remote)")
plt.ylabel("Number of Jobs")

plt.show()
```

```python
desc = df["salary_usd"].describe()
sns.barplot(x=desc.index, y=desc.values)
plt.title("Salary USD Descriptive Statistics")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["salary_usd"])

plt.title("Salary Distribution Boxplot")
plt.xlabel("Salary (USD)")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(x="company_size", y="salary_usd", data=df)

plt.title("Salary vs Company Size")
plt.xlabel("Company Size")
plt.ylabel("Salary (USD)")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(x="remote_ratio", y="salary_usd", data=df)

plt.title("Salary vs Remote Work Type")
plt.xlabel("Remote Ratio (0=Onsite, 50=Hybrid, 100=Remote)")
plt.ylabel("Salary (USD)")

plt.show()
```

```python
top_salary_roles = (
    df.groupby("job_title")["salary_usd"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_salary_roles.values, y=top_salary_roles.index)
plt.title("Top 10 Highest Paying AI Roles")
plt.xlabel("Mean Salary (USD)")
plt.ylabel("Job Title")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="remote_ratio",
    y="salary_usd",
    data=df
)

plt.title("Salary Distribution by Work Arrangement")
plt.xlabel("Remote Ratio (0 = Onsite, 50 = Hybrid, 100 = Fully Remote)")
plt.ylabel("Salary (USD)")

plt.show()
```

```python
df.groupby("remote_ratio")["salary_usd"].mean()
```

```python
remote_salaries = df.groupby("remote_ratio")["salary_usd"].mean()

plt.figure(figsize=(8, 5))
sns.barplot(x=remote_salaries.index, y=remote_salaries.values)
plt.title("Average Salary by Remote Work Ratio")
plt.xlabel("Remote Ratio (%)")
plt.ylabel("Mean Salary (USD)")
plt.show()
```

```python
country_salary = df.groupby("company_location")["salary_usd"].mean().reset_index()
country_salary
```

```python
top_country_salary = country_salary.sort_values(by="salary_usd", ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_country_salary, x="salary_usd", y="company_location")
plt.title("Top 10 Countries by Mean AI Salary")
plt.xlabel("Mean Salary (USD)")
plt.ylabel("Company Location")
plt.show()
```

```python
import plotly.express as px

fig = px.choropleth(
    country_salary,
    locations="company_location",
    locationmode="country names",
    color="salary_usd",
    color_continuous_scale="Viridis",
    title="Average AI Salary by Country",
)

fig.show()
```

```python
country_percent = df["company_location"].value_counts(normalize=True).head(10) * 100

plt.figure(figsize=(10,6))

sns.barplot(
    x=country_percent.values,
    y=country_percent.index
)

plt.xlabel("Percentage of AI Jobs (%)")
plt.ylabel("Country")
plt.title("Top Countries by Share of AI Jobs")

plt.show()
```

```python
remote_percent = df["remote_ratio"].value_counts(normalize=True) * 100

labels = ["Onsite (0)", "Hybrid (50)", "Remote (100)"]

plt.figure(figsize=(6,6))

plt.pie(
    remote_percent,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Distribution of Work Types in AI Jobs")

plt.show()
```

```python
exp_percent = df["experience_level"].value_counts(normalize=True) * 100

plt.figure(figsize=(6,6))

plt.pie(
    exp_percent,
    labels=exp_percent.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Distribution of Experience Levels in AI Jobs")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="experience_level",
    y="salary_usd",
    data=df
)

plt.title("Salary Distribution by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Salary (USD)")

plt.show()
```

```python
df["required_skills"] = df["required_skills"].str.split(",")
```

```python
skills_df = df.explode("required_skills")
```

```python
skills_df["required_skills"] = skills_df["required_skills"].str.strip()
```

```python
skill_salary = (
    skills_df
    .groupby("required_skills")["salary_usd"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=skill_salary.values,
    y=skill_salary.index
)

plt.title("Top 10 Highest Paying AI Skills")
plt.xlabel("Average Salary (USD)")
plt.ylabel("Skill")

plt.show()
```

```python
top_skills = skills_df["required_skills"].value_counts().head(10).index
```

```python
filtered = skills_df[skills_df["required_skills"].isin(top_skills)]
```

```python
skill_country_salary = (
    filtered
    .groupby(["company_location","required_skills"])["salary_usd"]
    .mean()
    .reset_index()
)
```

```python
plt.figure(figsize=(16, 8))
sns.heatmap(
    skills_df.groupby(["required_skills", "company_location"])["salary_usd"].mean().unstack(),
    cmap="viridis",
    annot=True,
    fmt=".0f",
    annot_kws={"size": 8},
    linewidths=0.5
).set(title="Average Salary by Skill and Country", xlabel="Country", ylabel="Skill")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

usa_df = skills_df[skills_df["company_location"] == "United States"]

plt.figure(figsize=(8, 6))
sns.heatmap(
    usa_df.groupby("required_skills")[["salary_usd"]].mean(),
    cmap="viridis",
    annot=True,
    fmt=".0f",
    linewidths=0.5
).set(title="Average Salary by Skill in United States", xlabel="Mean Salary (USD)", ylabel="Skill")
plt.show()
```
